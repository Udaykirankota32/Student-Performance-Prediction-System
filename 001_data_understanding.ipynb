```python
#loading the dataset
from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import numpy as np
import pandas as pd

file_path=Path(r"D:\students-Performance-Prediction-system\studentsDataFile\student\student-mat.csv")
df=pd.read_csv(file_path,sep=";")
# Easy (Minimal features)
easy_features = ["G1", "G2"]

# Realistic (Academic + behavioral)
realistic_features = [
    "G1", "G2", "studytime", "failures", "absences"
]

# Full (All useful features except target)
full_features = [col for col in df.columns if col != "G3"]
target="G3"

X_easy=df[easy_features]
X_realistic=df[realistic_features]
Y=df[target]
df_encoded = pd.get_dummies(df, drop_first=True)
y_full = df_encoded[target]

X_full = df_encoded.drop(columns=[target])

# Easy
X_train_easy, X_test_easy, y_train_easy, y_test_easy = train_test_split(
    X_easy,
    y,
    test_size=0.2,
    random_state=42
)

# Realistic
X_train_realistic, X_test_realistic, y_train_realistic, y_test_realistic = train_test_split(
    X_realistic,
    y,
    test_size=0.2,
    random_state=42
)

# Full

X_train_full, X_test_full, y_train_full, y_test_full = train_test_split(
    X_full,
    y_full,
    test_size=0.2,
    random_state=42
)


#LINEAR REGRESSION :

#EASY MODEL
lr_easy = LinearRegression()

lr_easy.fit(X_train_easy, y_train_easy)

#REALISITIC MODEL:
lr_realistic = LinearRegression()

lr_realistic.fit(X_train_realistic, y_train_realistic)

#FULL MODEL:
lr_full = LinearRegression()

lr_full.fit(X_train_full, y_train_full)





```

```python
#model evalution :
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
#Easy

r2_easy = r2_score(y_test_easy, y_pred_easy)
mae_easy = mean_absolute_error(y_test_easy, y_pred_easy)
rmse_easy = np.sqrt(mean_squared_error(y_test_easy, y_pred_easy))

#Realistic
r2_realistic = r2_score(y_test_realistic, y_pred_realistic)
mae_realistic = mean_absolute_error(y_test_realistic, y_pred_realistic)
rmse_realistic = np.sqrt(mean_squared_error(y_test_realistic, y_pred_realistic))


#full
r2_full = r2_score(y_test_full, y_pred_full)
mae_full = mean_absolute_error(y_test_full, y_pred_full)
rmse_full = np.sqrt(mean_squared_error(y_test_full, y_pred_full))
```

```python
#->The Easy model achieved the highest R² score (0.7946), indicating that it explains the largest proportion of variance in the final grade (G3).
#->The Easy model also produced the lowest MAE (1.2622) and lowest RMSE (2.0523), meaning its predictions are more accurate and have smaller errors compared to the Realistic and Full models.
#->Adding additional features in the Realistic and Full approaches did not improve the model's predictive performance.
#->The results suggest that G1 (first period grade) and G2 (second period grade) are the strongest predictors of G3 (final grade).
#->Including more features may have introduced noise, reducing the model's overall performance.
#->I am selecting the easy model to predict the results of the G3 as it's results are more accurate
```
