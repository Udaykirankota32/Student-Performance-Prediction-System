```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb

file_path=Path(r"D:\students-Performance-Prediction-system\studentsDataFile\student\student-mat.csv")

df=pd.read_csv(file_path,sep=";")


#Final Grade G3 distribution by age
plt.figure(figsize=(10,6))
sb.boxplot(data=df,x="age",y="G3")
plt.title("Final Grade (G3) Distribution by age")
plt.ylabel("final Grade (G3)")
plt.xlabel("Age")
plt.show()

print("\n Age category count and avg of final Grade G3:\n",df.groupby("age")["G3"].agg(["count", "mean"]).sort_values("mean", ascending=False).head(10))
```

```python
#From the boxplot, age 20 and 21 appear to have higher final grades. However, these age groups have very few students (only 1–3), so the result is not reliable.

#Most students are between 15–18 years, and among them, younger students (especially 15–16) show slightly better performance.
```

```python
#final Grade G3 distribution by study time
plt.figure(figsize=(10,6))
sb.boxplot(data=df, x="studytime", y="G3")
plt.title("Final Grade (G3) Distribution by Studytime")
plt.xlabel("Studytime")
plt.ylabel("Final Grade (G3)")
plt.show()
print("\nAverage G3 on Studytime:",df.groupby("studytime")["G3"].mean())
```

```python
#observation 
#The students who study  more that 10 hr and students with study time range 10-14 have high Mean G3
```

```python
#final Grade G3 distribution by no of failures
plt.figure(figsize=(10,6))
sb.boxplot(data=df, x="failures", y="G3")
plt.title("Final Grade (G3) Distribution by Failures")
plt.xlabel("Failures")
plt.ylabel("Final Grade (G3)")
plt.show()
print("\nAverage G3 on failures:",df.groupby("failures")["G3"].mean())
```

```python
#observation:
#-students with zero failure or minimum failures has higher avg final Grade (G3)
```

```python
#final Grade G3 distribution by FreeTime
plt.figure(figsize=(10,6))
sb.boxplot(data=df, x="freetime", y="G3")
plt.title("Final Grade (G3) Distribution by Studytime")
plt.xlabel("FreeTime")
plt.ylabel("Final Grade (G3)")
plt.show()
print("\n Freetime count and avg of final Grade G3:\n",df.groupby("freetime")["G3"].agg(["count", "mean"]).sort_values("mean", ascending=False).head(10))
```

```python
#Students with low free time (freetime = 2) have the highest average final grade (11.56), based on 64 students.

#Students with very high free time (freetime = 5) also have a good average (11.30), based on 40 students.

#Students with average free time (freetime = 3) have the lowest average final grade.
```

```python
#final Grade G3 distribution by No of Absences
# Create groups
df["absences_group"] = pd.cut(df["absences"], bins=[-1, 0, 5, 10, 20, 100],
                              labels=["0", "1-5", "6-10", "11-20", "20+"])

plt.figure(figsize=(8,5))
sb.boxplot(data=df, x="absences_group", y="G3")
plt.title("Final Grade (G3) by Absences Group")
plt.show()
print("\n Absences count and avg of final Grade G3:\n",df.groupby("absences")["G3"].agg(["count", "mean"]).sort_values("mean", ascending=False).head(10))
```

```python
#The boxplot of G3 vs Absences is hard to interpret because many absence values have only 1 student. Overall, students with fewer absences tend to have better and more stable final grades. Very high absences are rare.
```
