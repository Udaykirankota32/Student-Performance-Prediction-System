```python
#Data loading code
from pathlib import Path
import pandas as pd
import numpy as np

file_path=Path( r"D:\students-Performance-Prediction-system\studentsDataFile\student\student-mat.csv")
print(file_path.exists())
```

```python
df=pd.read_csv(file_path,sep=";")
display(df.head())
display(df.tail())
display(df.shape)
display(df.size)
display(df.info())
display(df.describe())
```

```python
#EDA ANALYSIS
print("\nG1 statistical data",df["G1"].describe())
print("\nG2 statistical data",df["G2"].describe())
print("\nG3 statistical data",df["G3"].describe())


```

```python
#univariate analysis
print("\nfinal Grade G3 statistics:",df["G3"].describe())
print("\nschools count:",df["school"].value_counts())
print("\nstudents Gender count:",df["sex"].value_counts())
print("\nstudent Age category count :",df["age"].value_counts())
print("\nstudent studytime:",df["studytime"].value_counts())
print("\nstudent failures:",df["failures"].value_counts())
print("\nstudent freetime:",df["freetime"].value_counts())
print("\nstudent absences:",df["absences"].value_counts())
print("\nAvg final Grade G3:",df["G3"].mean())
print("\nstandard Deviation of final Grade G3:",df["G3"].std())


```

```python
#Bivariate Analysis on G3 relationship with other factors

print("\nschool's students Final Grade G3 avg",df.groupby("school")["G3"].mean())
print("\nAverage G3 on Gender",df.groupby("sex")["G3"].mean())
print("\nAverage G3 on age",df.groupby("age")["G3"].mean())
print("\n Age category count and avg of final Grade G3:\n",df.groupby("age")["G3"].agg(["count", "mean"]).sort_values("mean", ascending=False).head(10))
print("\nAverage G3 on Studytime:",df.groupby("studytime")["G3"].mean())
print("\nAverage G3 on failures:",df.groupby("failures")["G3"].mean())
print("\nAverage G3 on Freetime:",df.groupby("freetime")["G3"].mean())
print("\n Freetime count and avg of final Grade G3:\n",df.groupby("freetime")["G3"].agg(["count", "mean"]).sort_values("mean", ascending=False).head(10))
print("\nAverage G3 on absences:",df.groupby("absences")["G3"].mean())
print("\n absences count and avg of final Grade G3:\n",df.groupby("absences")["G3"].agg(["count", "mean"]).sort_values("mean", ascending=False).head(10))
```

```python
##observation:
#->GP School has higher final Grade Avg compare to MP
#->Males have more Final grade avg compare to Female
#->Students with age 21 recorded highest final Grade Avg=14.0 but Younger students (especially age 15) have slightly better average final grades compared to older students in this dataset. The highest average at age 20 is based on very few students, so it should not be considered strongly.
#->students who studies 5 to 10 hours(3rd category) have highest final grade avg=11.4
#->students with zero failures have highest final Grade G3 avg=11.23
#->students with low free time unexpectedly have more final grade G3 avg=11.56 followed by students with high free time with avg=11.3
#->At first look, students with 24 absences showed the highest average final grade (18.0). However, this is misleading because only one student had 24 absences.When we consider groups with a reasonable number of students, students with 2 absences have a relatively high average final grade (approximately 11.91).
```

```python
#reloaded the data
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sb


file_path=Path( r"D:\students-Performance-Prediction-system\studentsDataFile\student\student-mat.csv")
df=pd.read_csv(file_path,sep=";")

#visualization
plt.figure(figsize=(10,6))
sb.histplot(data=df,x="G3", bins=20,kde=True)
plt.title("Distribution of Final Grade (G3)")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")
plt.show()
```

```python
###visual Exploriatory Analysis
```

```python

##visualizing all the univariate
#schools categorical students data
plt.figure(figsize=(10,5))
sb.countplot(data=df,x="school")
plt.title("Distribution of the students across different schools")
plt.xlabel("schools")
plt.ylabel("students")
plt.show()

```

```python
#distribution of students across gender
plt.figure(figsize=(10,6))
sb.countplot(data=df,x="sex")
plt.title("Distribution of students across gender")
plt.xlabel("Gender")
plt.ylabel("Students")
plt.show()
```

```python
#distribution of students across age
plt.figure(figsize=(10,6))
sb.histplot(data=df,x="age",bins=10,kde=True)
plt.title("Distribution of students across age")
plt.xlabel("Age")
plt.ylabel("Students")
plt.show()
```

```python
#distribution of students vs studytime
plt.figure(figsize=(10,6))
sb.countplot(data=df,x="studytime")
plt.title("Distribution of students across Studytime")
plt.xlabel("studytime")
plt.ylabel("Students")
plt.show()
```

```python
#distribution of students vs failures
plt.figure(figsize=(10,6))
sb.countplot(data=df,x="studytime")
plt.title("Distribution of students vs Failures")
plt.xlabel("Failures")
plt.ylabel("Students")
plt.show()
```

```python
#distribution of students vs Freetime
plt.figure(figsize=(10,6))
sb.countplot(data=df,x="freetime")
plt.title("Distribution of students across Freetime")
plt.xlabel("freetime")
plt.ylabel("Students")
plt.show()
```

```python
#distribution of students vs Absences
plt.figure(figsize=(12,8))
sb.histplot(data=df,x="absences",bins=94,kde=True)
plt.title("Distribution of students vs Absences")
plt.xlabel("Absences")
plt.ylabel("Students")
plt.show()
```

```python
#distribution of students across Final Grade G3 
plt.figure(figsize=(10,6))
sb.boxplot(data=df,x="G3")
plt.title("Distribution of students across  Final Grade G3 ")
plt.xlabel(" Final Grade G3 average")
plt.ylabel("Students")
plt.show()
```

```python
### Univariate Analysis – Observations

##Final Grade (G3):
#- Average final grade is approximately **10.42**
#- Median is **11**
#- Grades range from 0 to 20
#- Most students scored between 8 and 14
#- Standard deviation is 4.58, which shows moderate spread in scores

##School:
#- Majority of students are from **GP school** (349 students)
#- Only 46 students are from MS school

#Gender:
#- Female students: 208
#- Male students: 187
#- Dataset is almost balanced in terms of gender

#Age:
#- Most students are between 15 to 18 years
#- Highest number of students are 16 years old (104 students)
#- Very few students are above 19 years

##Study Time:
# ---Most students study 2–5 hours per week (studytime = 2)
# --Very few students study more than 10 hours (studytime = 4)

##Failures:
#- Most students have **0 failures** (312 students)
#- Only a small number of students have 2 or 3 failures

##Free Time:
#- Most students have average free time (freetime = 3)
#- Very few students have very low free time

##Absences:
#- Many students have 0 absences (115 students)
#- Number of absences is mostly low
#- Very high absences (above 20) are rare

```
